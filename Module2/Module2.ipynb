```python
Question 1: Character Frequency Counter
Write a function:

that returns frequency of each character in a string.

Example 1
Input: "apple"
Output: {
    'a':1,
    'p':2,
    'l':1,
    'e':1
}
```

```python
def char_freq_counter(string):

    d={}
    for i in string:
        if i in d:
            d[i]+=1
        else:
            d[i]=1

    return d
```

```python
string='apple'

for i in string:
    print(i)
```

```python
#Without Error Handling
#Type a text, create intentional error

number = int(input("Enter a number: "))
print(number)
```

```python
#Basic Try Except
try:
    number = int(input("Enter a number: "))
    print(number)
except:
    print("Invalid input")
```

```python
result = 10/0
```

```python
#Specific Exception
try:
    result = 10/0
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

# Part 6: Mini Exercise

### Final Mini Project

```python
import json

def recommend_product(category):
    
    products = {
        "electronics": ["Laptop", "Phone"],
        "fashion": ["Shoes", "Shirt"],
        "books": ["Python Book", "AI Book"]
    }
    
    if category not in products:
        raise ValueError("Please type a valid category")
    
    return json.dumps({
        "category": category,
        "recommendations": products[category]
    })
```

```python
type(recommend_product("electronics"))
```

```python
print(recommend_product('electronics'))
```

```python
#Test Invalid Input
print(recommend_product('food'))
```

```python
# Invalid input with error handling
try:
    print(recommend_product("food"))
except ValueError as e:
    print(e)
```
