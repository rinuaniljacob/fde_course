```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()

client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY")
)

print("API Loaded Successfully")
```

# Available Departments

Our company has four departments.

Each department has its own email address.

```python
departments = {
    "Sales": "sales@company.com",
    "Support": "support@company.com",
    "Marketing": "marketing@company.com",
    "Tech": "tech@company.com"
}
```

# Email Received From Customer

Students can change this email and test different scenarios.

```python
email_text = """
Hello Team,

I'm facing an issue with the payment gateway. My order ID is 123.

Thanks
"""
```

# Ask GPT to Classify the Email

The model must choose one department only.

```python
prompt = f"""
You are an email routing assistant.

Choose ONLY one department.

Departments:

1. Sales
   - Product demos
   - Pricing
   - New customers

2. Support
   - Complaints
   - Refunds
   - Order issues

3. Marketing
   - Partnerships
   - Advertising
   - Promotions

4. Tech
   - Login issues
   - Bugs
   - API issues

Return ONLY the department name.

Email:

{email_text}
"""
```

```python
prompt
```

# Send Email to GPT

GPT reads the email and selects the department.

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0
)

department = response.choices[0].message.content.strip()

print("Predicted Department:")
print(department)
```

```python
department_email = departments.get(
    department,
    "support@company.com" # Default value
)

print("Route To:")
print(department_email)
```
